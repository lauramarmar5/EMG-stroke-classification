```python
def plot_emg(i, musculo, estado):
  signal = pd.read_csv(f'/content/database/{estado}/{musculo}/MVC {musculo}_{i}.csv', skiprows=11)
  amplitud = signal.iloc[:,1]*1e3 #pasar a milivoltios
  t = signal.iloc[:,0]
  plt.plot(t, amplitud)
  plt.xlabel('Tiempo (s)')
  plt.ylabel('Amplitud (mV)')
  plt.title('Registro de amplitud de EMG')
  plt.show()

def saca_metricas(i, estado, musculo):
  signal = pd.read_csv(f'/content/database/{estado}/{musculo}/MVC {musculo}_{i}.csv', skiprows=11)
  amplitud = signal.iloc[:,1]*1e3 #pasar a milivoltios

  # calculo de metricas
  max = amplitud.max()
  mean = amplitud.mean()
  std = amplitud.std()
  median = amplitud.median()
  ssi = np.sum(amplitud**2)
  rms = np.sqrt(np.mean(amplitud**2))

  if estado == 'enfermo':
    caso_control = 1
  if estado == 'sano':
    caso_control = 0
  return max, mean, std, median, ssi, rms, caso_control

def crea_df(musculo):
  datos = []

  for i in range(1,13):
    metricas = saca_metricas(i, 'sano', musculo)
    datos.append(metricas)

  for i in range(1,11):
    metricas = saca_metricas(i, 'enfermo', musculo)
    datos.append(metricas)

  df = pd.DataFrame(datos, columns = ['Max', 'Mean',, 'Median', 'Std','SSI', 'RMS', 'Group'])
  df = df.sample(frac=1).reset_index(drop=True)
  df.to_csv(f'{musculo}')

  return df
```
